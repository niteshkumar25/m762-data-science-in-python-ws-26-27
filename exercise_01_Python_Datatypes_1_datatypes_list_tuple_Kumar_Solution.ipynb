### 7. Perform a comparsion of two float values x and y! (Larger, equal, lower)

```python
x = 4.0
y = 6.0

print("x > y :", x > y)
print("x < y :", x < y)
print("x == y:", x == y)
print("x >= y:", x >= y)
print("x <= y:", x <= y)
print("x != y:", x != y)
```

 ### for C = 4 and D = 5

### 11. What is the difference between a list and a tuple?


A list is mutable (can be changed after creation, written with []), while a tuple is immutable (cannot be changed, written with ()).

A = [1, 2, 3]   # list
A[0] = 99       # ✅ works → A = [99, 2, 3]

B = (1, 2, 3)   # tuple
B[0] = 99       # ❌ TypeError: 'tuple' object does not support item assignment
