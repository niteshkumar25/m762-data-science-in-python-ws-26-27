## 1. enumerate


- when you iterating over a sequence and want to keep track of the index of the current item

            # common way:
            i = 0
            for value in collection:
                # do something with i and value
                i += 1  

```python
# enumerate returns (index, value) pairs
collection = ['apple', 'banana', 'cherry']

for i, value in enumerate(collection):
    print(i, value)
```

```python
# enumerate example

some_list = ['name', 'region', 'energy']
mapping = {} # empty dict
for i, v in enumerate(some_list):
    mapping[i] = v    # try to change i and v  --> what happens?
mapping
```

# Now your task:

Iterate trough the values of your self-created *MyDict* and sort the elements of each list.

Goal: Two new dictionaries, one of them with forward sorted lists as values and  one of them with reverse sorted lists as values 


```python
# Exercise 2: sort each list in MyDict
SortedDict = {}
ReverseSortedDict = {}

for key, values in MyDict.items():
    SortedDict[key] = sorted(values)
    ReverseSortedDict[key] = sorted(values, reverse=True)

print("Forward sorted:", SortedDict)
print("Reverse sorted:", ReverseSortedDict)
```

## 3. zip

- zip pairs up (verbindet) the elements of a number of lists, tuples or other sequences to create a list of tuples

```python
seq1 = ['foo', 'bar', 'egg']
seq2 = ['one', 'two', 'three']
zipped = zip(seq1, seq2)
zipped
```

```python
list(zipped)
```

```python
for i in zip(seq1, seq2):

    print(i)
```

### A common use of zip is simultaneously iterating over multiple sequences, possibly also combined with enumerate:

```python
for i, (a, b) in enumerate(zip(seq1, seq2)):
        print('{0}: {2}, {1}'.format(i, a, b))
```

### or use it for clever unzip a sequence

```python
participants = [('Tom', 'Miller'), ('Roger', 'Clemens'), ('Max', 'Schubert')]
first_names, last_names = zip(*participants)  # * stands for unzip
```

```python
first_names
```

```python
last_names
```

```python
first_names = []
last_names = []

for i in participants:
    print(i)
    first_names.append(i[0])
    last_names.append(i[1])

first_names, last_names
    
```
