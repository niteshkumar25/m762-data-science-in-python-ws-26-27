1. Create a pandas series ***GZ_1*** from all even numbers between 0 and 20 (the limits should be included).

```python
import pandas as pd

GZ_1 = pd.Series(range(0, 21, 2))
print(GZ_1)
```

2. Index from ***GZ_1*** every 2nd element and write it into a pandas series ***GZ_2***.

```python
GZ_2 = GZ_1[::2]
print(GZ_2)
```

3. Change the index from ***GZ_2*** to letters from ***a*** to ...

```python
GZ_2.index = list('abcdef')
print(GZ_2)
```

4. Write the last 3 entries in a new pandas series ***GZ_3***.

```python
GZ_3 = GZ_2[-3:]
print(GZ_3)
```
