### 13. Create a list with 3  weekday names. Go through this list with a **for loop** and print 'weekend' if the weekday name is 'Saturday' or 'Sunday', else print 'working day'.

```python
weekdays = ["Monday", "Saturday", "Wednesday"]
for day in weekdays:
    if day in ["Saturday", "Sunday"]:
        print(day, "- weekend")
    else:
        print(day, "- working day")
```

### 15. Create a class object "wind turbine". Give this objects some simple propreties and methods.

```python
class WindTurbine:
    def __init__(self, location, capacity_kw, is_running=False):
        self.location = location
        self.capacity_kw = capacity_kw
        self.is_running = is_running

    def start(self):
        self.is_running = True
        print(f"Wind turbine at {self.location} is now running.")

    def stop(self):
        self.is_running = False
        print(f"Wind turbine at {self.location} has stopped.")

    def show_status(self):
        status = "running" if self.is_running else "stopped"
        print(f"Location: {self.location}")
        print(f"Capacity: {self.capacity_kw} kW")
        print(f"Status: {status}")

# Create an object and test its methods
turbine = WindTurbine("Germany", 2500)
turbine.show_status()
turbine.start()
turbine.show_status()
turbine.stop()
```
